#### <span class="girk">use of isin()</span>

````python
def team1vsteam2(str1,str2):
    
    df_new = ipl[((ipl['Team1'].isin([str1,str2])) & (ipl['Team2'].isin([str1,str2])))]
    
    print(f'total matches played between {str1} and {str2} is {df_new.shape[0]}')
    print()
    
    winners = df_new['WinningTeam'].value_counts().reset_index()
    
    print(f'{winners.iloc[:,0][0]} has won for {winners.iloc[:,1][0]} and {winners.iloc[:,0][0]} lost for {winners.iloc[:,1][1]} matches')
    print()
    
    
    pom = df_new['Player_of_Match'].value_counts()[(df_new['Player_of_Match'].value_counts()==df_new['Player_of_Match'].value_counts().max())]
    
    pom1 = pom.reset_index()
    
    
    print(f'max man of the match is {pom1.iloc[:,0][0]} for {int(pom1.iloc[:,1].values)} matches')
````

````python
team1vsteam2('Kolkata Knight Riders','Chennai Super Kings')
````

````python
ipl.replace('Delhi Daredevils','Delhi Capitals',inplace=True)
ipl.replace('Kings XI Punjab','Punjab Kings',inplace=True)
ipl.replace('Rising Pune Supergiants','Rising Pune Supergiant',inplace=True)
````

````python
def point_table(season):
    
    ipl_new = ipl[ipl['Season'] == season]
    total_teams = pd.concat((ipl_new['Team1'],ipl_new['Team2']))
    teams = [i for i in set(total_teams)]
    
    data = pd.DataFrame(columns=['Team_name','Total_Matches_Played','Total_Won','No_results','Points'])
    
    for key,i in enumerate(teams):
        
        total_matches_played = (total_teams == i).sum()
        
        matches_won = len(ipl_new[ipl_new['WinningTeam'] == i])
        
        no_result = ipl_new[ipl_new['WonBy']=='NoResults'][['Team1','Team2']].apply(lambda x : x==i).sum().sum()
        
        points = (matches_won*2) + no_result
        
        info = [i,total_matches_played,matches_won,no_result,points]
        
        data.loc[key] = info
    
    data.set_index('Team_name',inplace=True)
    
    return data.sort_values(['Points','Total_Won'],ascending=False).head(2)
    
````

````python
def batsman_stats(player_name):
    
    sub1 = sub2[((sub2['batter'] == player_name) | ((sub2['player_out']==player_name)))]
    # print('sub1 is the enitire dataset of the player',sub1.shape)
    
    sub1['actual_runs'] = sub1.apply(lambda row: 0 if ((row['isWicketDelivery']==1)&(row['batter'] != player_name)) else row['batsman_run'], axis=1)
    
    sub1['actual_out'] = sub1.apply(lambda row: 1 if ((row['player_out'] == player_name)) else 0, axis=1)
    
    sub1['actual_delivery'] = sub1.apply(lambda row: 1 if ((row['batter']==player_name)&(row['extra_type'] != 'wides')) else 0, axis=1)
    
    
    df1 = sub1.groupby(['Season','ID']).agg(total_runs_scored = ('actual_runs','sum'))
    
    m = df1.groupby(['Season'])['total_runs_scored'].agg(['max'])
    m['innings'] = df1.reset_index().groupby('Season')['Season'].agg('count')
    data1 = m.rename(columns={'max':'highest_score_season'})
    
    
    df1 = sub1.groupby(['Season']).agg(
    #total_innings_played = ('innings','count'),
    total_runs_scored = ('actual_runs','sum'),
    total_out = ('actual_out','sum'),
    total_balls_played = ('actual_delivery','sum'))
    
    df1['avg'] = round((df1['total_runs_scored']/df1['total_out']),2)
    df1['strike_rate'] = round(((df1['total_runs_scored']/df1['total_balls_played'])*100),2)
    
    result = df1.merge(data1,on='Season').rename(columns={'total_runs_scored_x':'total_runs_scored','total_runs_scored_y':'max_score','ID':'innings'})
    results = result[['innings','total_out','total_runs_scored','highest_score_season','total_balls_played','strike_rate','avg']][::-1]
    
    
    
    enitre_run, entire_innings, entire_balls = results['total_runs_scored'].sum(),results['innings'].sum(),results['total_balls_played'].sum()
    
    entire_highest_run, entire_not_out = results['highest_score_season'].max() , (results['innings'].sum() - results['total_out'].sum())
    
    entire_avg = round((results['total_runs_scored'].sum()/results['total_out'].sum()),2)
    
    print(f'total runs scored by {player_name} : {enitre_run} runs in {entire_innings} innings facing {entire_balls} balls')
    print()
    print(f'highest score by {player_name} is {entire_highest_run} runs and was notout for {entire_not_out} times with an average of {entire_avg} runs')
    
    
    
    return results
 
````
