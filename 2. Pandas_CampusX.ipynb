## Rank (Series)

The `rank()` method in pandas is used to assign ranks to the values in a Series. The rank of a value is determined by its position relative to other values in the Series. By default, ranks are assigned starting from 1, and the smallest value gets the rank 1.

### Key Parameters of `rank()`:

- **`method`**: Specifies the method used to assign ranks when there are ties (i.e., when two or more values are equal).
  - `'average'` (default): Assigns the average rank to tied values.
  - `'min'`: Assigns the minimum rank to tied values.
  - `'max'`: Assigns the maximum rank to tied values.
  - `'first'`: Assigns ranks in the order the values appear in the Series.
  - `'dense'`: Like `'min'`, but ranks are always consecutive.



- **`ascending`**: If `True`, ranks are assigned in ascending order (default). If `False`, ranks are assigned in descending order.



- **`na_option`**: Specifies how to handle `NaN` values.
  - `'keep'` (default): Leaves `NaN` values in place.
  - `'top'`: Assigns `NaN` the smallest rank.
  - `'bottom'`: Assigns `NaN` the largest rank.

## Difference between unique and nunique

- **`unique()`**: Returns an array of the unique values in a Series or DataFrame column. It includes all distinct values without counting how often they appear.


- **`nunique()`**: Returns the __number of unique values__ in a Series or DataFrame column. It's a count of distinct values, not the values themselves.

### Example:

- **`unique()`**: If you have a Series `[1, 2, 2, 3]`, `unique()` will return `[1, 2, 3]`.

- **`nunique()`**: For the same Series, `nunique()` will return `3`.

## <span class="girk">replacing values using np.where according to a condition</span>

`np.where()` is used to conditionally assign values in a DataFrame or Series based on a condition. 


It's a function when you want to apply logic that selects between two choices depending on whether a condition is met or not.

### Basic Syntax:
```python
np.where(condition, value_if_true, value_if_false)
```

- `condition`: A condition that returns a boolean Series or array (True/False values).


- `value_if_true`: The value to assign if the condition is `True`.


- `value_if_false`: The value to assign if the condition is `False`.

### Example 1: Assigning Values to a DataFrame Column
Let's say you have a DataFrame with a column named `scores`, and you want to create a new column `pass_fail` that indicates whether each score is a pass (greater than or equal to 50) or fail (less than 50).

```python
import pandas as pd
import numpy as np

# Sample DataFrame
df = pd.DataFrame({
    'scores': [45, 67, 89, 23, 50]
})

# Use np.where to create a new column based on the condition
df['pass_fail'] = np.where(df['scores'] >= 50, 'Pass', 'Fail')

print(df)
```

**Output:**
```
   scores pass_fail
0      45      Fail
1      67      Pass
2      89      Pass
3      23      Fail
4      50      Pass
```

## Difference between concat, merge and join in pandas

In pandas, concat, merge, and join are used to combine data frames. However, they differ in their functionality and the way they combine data frames.

- __concat :__ concat is used to stack one data frame on top of another or to concatenate data frames along an axis. It can concatenate two or more data frames either vertically axis 0 (i.e., along rows) or horizontally (i.e., along columns). 


- __merge :__ merge is used to merge two data frames based on a __common column or index__. It is similar to a join operation in SQL.  The how parameter is used to specify the type of join to be performed (inner, outer, left or right join).


<span class="girk">***In summary, concat is used to concatenate data frames along an axis, merge is used to merge data frames based on a common column or index, and join is used to join data frames based on their indexes.***</span>

### <span class="girk">Why do we need pandas timestamp we we already have python's datetime ?</span> 

- syntax wise datetime is very convenient


- But the performance takes a hit while working with huge data. List vs Numpy Array


- The weaknesses of Python's datetime format inspired the NumPy team to add a set of native time series data type to NumPy.


- The datetime64 dtype encodes dates as 64-bit integers, and thus allows arrays of dates to be represented very compactly.

````python
import numpy as np
date = np.array('2015-07-04', dtype=np.datetime64)
date
````

#### vector operations on date

````python
date + np.arange(12)
````

- Because of the uniform type in NumPy datetime64 arrays, this type of operation can be accomplished much more quickly than if we were working directly with Python's datetime objects, especially as arrays get large 



- <span class="girk">Pandas Timestamp object combines the ease-of-use of python datetime with the efficient storage and vectorized interface of numpy.datetime64</span>



- From a group of these Timestamp objects, Pandas can construct a DatetimeIndex that can be used to index data in a Series or DataFrame
