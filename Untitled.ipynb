```python
# number of movies in each genres which starts with 'A'

def stratswith_a(group):
    return group['Series_Title'].str.startswith('A').sum()
    
```

```python
genres = movies.groupby('Genre')
genres
```

```python
# find ranking of each movie in their group according to IMDB score :

def ranking(group):
    group['ranking'] = group['IMDB_Rating'].rank(ascending=False)
    return group
```

```python
genre.apply(ranking)
```

```python
def normalized_ranking(group):
    group['normalized_rating'] = (group['IMDB_Rating'] - group['IMDB_Rating'].min())/(group['IMDB_Rating'].max() - group['IMDB_Rating'].min())
    return group
```

```python
genres.apply(normalized_ranking,include_groups=False)
```
