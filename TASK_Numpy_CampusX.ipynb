````python
def create_array(a, b):
    # Create a 2D array of zeros with shape (a,b)
    arr = np.zeros((a, b))

    # Set the first and last rows to 1
    arr[0, :] = 1
    arr[-1, :] = 1

    # Set the first and last columns to 1
    arr[:, 0] = 1
    arr[:, -1] = 1

    return arr
````

### `Q-20:` <span class="girk">Create the following pattern without hardcoding. Use only numpy functions and the below input array a.</span>
```
# Input: a = np.array([1,2,3])
# Output: array([1, 1, 1, 2, 2, 2, 3, 3, 3, 1, 2, 3, 1, 2, 3, 1, 2, 3])
```

````python
def pattern(a,times):
    
    a1 = np.repeat(a, times)
    a2 = np.tile(a,times)
    
    return np.concatenate((a1,a2))
````

````python
a = np.array([1,2,3])
a
````

````python
pattern(a,3)
````

A Cauchy matrix is a mathematical matrix in which each element (i, j) is defined as 1 / (xi - yj), where xi and yj are the ith element of vector x and the jth element of vector y, respectively.
